## 2. Decoupling the Five Core Statistical Performance Concepts
To ensure complete methodological transparency, our audit strictly separates:
1. **Actual Pressure Performance ($SR_{\text{press}}$ / $Econ_{\text{lev}}$):** Observable scoring or economy under acute pressure.
2. **Normal Career Baseline ($SR_{\text{normal}}$ / $Econ_{\text{normal}}$):** Career performance during low-pressure phases ($\Psi < 0.45$, $\Lambda < 0.40$).
3. **Context-Expected Par ($SR_{\text{expected}}$ / $Econ_{\text{expected}}$):** Situational par derived from historical over-by-over, venue-specific baselines.
4. **Context-Adjusted Delta ($\Delta SR = SR_{\text{press}} - SR_{\text{expected}}$, $\Delta Econ = Econ_{\text{lev}} - Econ_{\text{expected}}$):** Pure situational elevation above match par.
5. **Bayesian Posterior Estimate ($\theta_{\text{Bayes}}$):** Shrunk toward empirical population priors ($K_{\text{bat}}=45$, $K_{\text{bowl}}=55$) to prevent small-sample misallocation.
