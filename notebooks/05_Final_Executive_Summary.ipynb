## 4. Empirical Bayesian Shrinkage & Credible Intervals
Maximum Likelihood Estimators (MLE) are prone to extreme sample volatility in T20 cricket. The Bayesian engine applies empirical prior shrinkage:
- **Batting Prior**: $\mu_{\text{prior, SR}} = 135.20$, $K_{\text{bat}} = 45.0$
- **Bowling Prior**: $\mu_{\text{prior, Econ}} = 8.42$, $K_{\text{bowl}} = 55.0$

$$w_i = \frac{N_i}{N_i + K}, \quad \theta_{\text{Bayes}} = w_i \theta_{\text{Observed}} + (1 - w_i) \mu_{\text{Prior}}$$

This guarantees that emerging players like Naman Dhir (91 balls) receive appropriate credit without exposing the franchise purse to small-sample illusions, while marquee players like Tim David (207 balls) and Jasprit Bumrah (763 balls) retain high-certainty ratings with razor-thin credible intervals.
